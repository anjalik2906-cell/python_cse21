```python
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun']
sales = [100, 120, 90, 150, 200, 170]
plt.plot(months, sales)
```

```python
#change line style
plt.plot(
    months,
    sales,
    color='green',
    linestyle='dashed',
    linewidth=2,
    marker='o',
)
```

```python
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']
sales_2025 = [100, 120, 90, 150, 200, 170, 180, 220, 250, 300, 280, 320]
sales_2026 = [110, 130, 100, 160, 210, 180, 190, 230, 260, 310, 290, 330]
plt.plot(months, sales_2025, label='2025', color='pink', linestyle='dashed', linewidth=2, marker='o')
plt.plot(months, sales_2026, label='2026', color='purple', linestyle='dashed', linewidth=2, marker='o') 
```
