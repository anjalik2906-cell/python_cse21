```python
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']
sales_2025 = [100, 120, 90, 150, 200, 170, 180, 220, 250, 300, 280, 320]
sales_2026 = [110, 130, 100, 160, 210, 180, 190, 230, 260, 310, 290, 330]
plt.plot(months, sales_2025, label='2025', color='pink', linestyle='dashed', linewidth=2, marker='o')
plt.plot(months, sales_2026, label='2026', color='purple', linestyle='dashed', linewidth=2, marker='o') 
```

```python
import matplotlib.pyplot as plt
student =["A","B","C","D","E"]
python=[80,90,70,60,50]
java=[70,60,80,90,50]
plt.bar(student, python, label="Python", color='pink')
plt.bar(student, java, label="Java", color='lightgreen', bottom=python)
```

```python
import matplotlib.pyplot as plt
student=['alice','bob','charlie','david','eve']
hours=[10,15,12,8,20]
explode=[0,0.1,0,0,0]
sycolors=['pink','lightgreen','lightblue','violet','purple']
plt.pie(hours, labels=student, explode=explode, colors=sycolors, autopct='%1.1f%%', shadow=True, startangle=90)
plt.title('Hours spent on coding by students')
```

```python
import pandas as pd
```

```python
data=pd.read_csv(r"C:\Users\khush\Downloads\Titanic-Dataset.csv")
```

```python
plt.hist(data["Age"].dropna(), bins=20)
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.scatter(data['Age'], data['Fare'], color='violet', alpha=0.5)
plt.title("Fare vs Age of Passengers")
plt.xlabel("Age")
plt.ylabel("Fare")
```

```python
age_data = [
    data[data["Pclass"] == 1]["Age"].dropna(),
    data[data["Pclass"] == 2]["Age"].dropna(),
    data[data["Pclass"] == 3]["Age"].dropna()
]
plt.boxplot(age_data)
plt.title("Age Distribution by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Age")
plt.show()
```
